```python
import re
import requests
from bs4 import BeautifulSoup

HEADERS = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"}

def to_num(text):
    m = re.search(r"[\d,]+(?:\.\d+)?", text or "")
    return float(m.group().replace(",", "")) if m else None

def get_num(soup, selector):
    tag = soup.select_one(selector)
    return to_num(tag.get_text()) if tag else None

def crawl_book(goods_no):
    res = requests.get(f"https://www.yes24.com/product/goods/{goods_no}", headers=HEADERS)
    res.raise_for_status()
    soup = BeautifulSoup(res.text, "html.parser")
    title = soup.select_one("h2.gd_name")
    isbn = soup.select_one('meta[property="books:isbn"]')
    return {
        "goods_no": goods_no,
        "isbn": isbn["content"] if isbn else None,
        "title": title.get_text(strip=True) if title else None,
        "rating": get_num(soup, "span.gd_rating em.yes_b"),
        "review_count": get_num(soup, "span.gd_reviewCount em.txC_blue"),
        "sales_index": get_num(soup, "span.gd_sellNum"),
    }

def show_book(b):
    fmt = lambda v, f: f.format(v) if v is not None else "-"
    print(f"제목      : {b['title']}")
    print(f"평균별점  : {fmt(b['rating'], '{:.1f}')} / 10")
    print(f"리뷰건수  : {fmt(b['review_count'], '{:,.0f}')}건")
    print(f"판매지수  : {fmt(b['sales_index'], '{:,.0f}')}")

book = crawl_book("139807878")
show_book(book)
```

```python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup

LIST_URL = "https://www.yes24.com/product/category/bestseller"
HEADERS = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"}

res = requests.get(LIST_URL, headers=HEADERS,
                   params={"categoryNumber": "001", "pageNumber": 1, "pageSize": 120})
res.raise_for_status()
items = BeautifulSoup(res.text, "html.parser").select("ul#yesBestList > li")

rows = []
for li in items:
    name = li.select_one("a.gd_name")
    goods_no = li.get("data-goods-no")
    rows.append({
        "rank": int(li.select_one("em.ico.rank").get_text(strip=True)),
        "title": name.get_text(strip=True) if name else None,
        "goods_no": goods_no,
        "url": f"https://www.yes24.com/product/goods/{goods_no}",
    })

df_list = pd.DataFrame(rows)
print(df_list.shape)
df_list.head()
```

```python
import re, time
import requests
import pandas as pd
from bs4 import BeautifulSoup

def to_num(text):
    m = re.search(r"[\d,]+(?:\.\d+)?", text or "")
    return float(m.group().replace(",", "")) if m else None

def get_num(soup, selector):
    tag = soup.select_one(selector)
    return to_num(tag.get_text()) if tag else None

def crawl_book(goods_no):
    url = f"https://www.yes24.com/product/goods/{goods_no}"
    res = requests.get(url, headers=HEADERS)
    res.raise_for_status()
    soup = BeautifulSoup(res.text, "html.parser")
    title = soup.select_one("h2.gd_name")
    isbn = soup.select_one('meta[property="books:isbn"]')
    return {
        "goods_no": goods_no,
        "url": url,
        "isbn": isbn["content"] if isbn else None,
        "title": title.get_text(strip=True) if title else None,
        "rating": get_num(soup, "span.gd_rating em.yes_b"),
        "review_count": get_num(soup, "span.gd_reviewCount em.txC_blue"),
        "sales_index": get_num(soup, "span.gd_sellNum"),
    }

def show_book(b):
    fmt = lambda v, f: f.format(v) if v is not None else "-"
    print(f"제목      : {b['title']}")
    print(f"평균별점  : {fmt(b['rating'], '{:.1f}')} / 10")
    print(f"리뷰건수  : {fmt(b['review_count'], '{:,.0f}')}건")
    print(f"판매지수  : {fmt(b['sales_index'], '{:,.0f}')}")

def crawl_all(goods_nos, delay=1.0, verbose=True):
    results = []
    for i, no in enumerate(goods_nos, 1):
        try:
            book = crawl_book(no)
            results.append(book)
            if verbose:
                print(f"[{i}/{len(goods_nos)}]")
                show_book(book)
                print("-" * 30)
        except Exception as e:
            print(f"[실패] {i}번 {no}: {e}")
        time.sleep(delay)
    return pd.DataFrame(results)
```

```python
df_ext = crawl_all(df_list["goods_no"], verbose=False)
print(df_ext.shape)
```

```python
# 1) 수집 성공/실패 개수
success = set(df_ext["goods_no"])
failed = df_list[~df_list["goods_no"].isin(success)]
print(f"성공 {len(success)}권 / 실패 {len(failed)}권")
print(failed[["rank", "title", "goods_no"]])      # 실패한 책 목록

# 2) 값이 비어 있는 책 (수집은 됐지만 추출 실패)
cols = ["rating", "review_count", "sales_index", "isbn", "title"]
print(df_ext[cols].isna().sum())
print(df_ext[df_ext[cols].isna().any(axis=1)][["goods_no", "title"] + cols[:3]])
```
