## Python Part1 - 리스트와 반복문을 이용한 데이터처리

> 여러 개의 데이터를 리스트에 저장하고 for 반복문을 이용하여 하나씩 처리합니다.  

1. 리스트 생성
2. 리스트의 데이터 순회
3. 조건에 맞는 데이터 찾기
4. 조건을 만족하는 데이터 개수 계산
5. 계산 결과 출력

> 필수 사용 - list / for / if

### Part 1. 리스트와 반복문을 이용한 데이터 처리

1. 리스트 생성  
- 필수 역량 `list`, `for`, `if`

```python
scores = [65, 80, 72, 90, 55, 85, 70]
```

2. 리스트의 데이터 순회  
- 각 점수를 하나씩 꺼내서 확인한다.

```python
for i in range(7):
    print(scores[i])
```

3. 조건에 맞는 데이터 찾기
- 70점 이상인 데이터를 찾아 출력한다.

```python
for score in scores:
    if score >= 70:
        print(score)
```

4. 조건을 만족하는 데이터 개수 계산.
- 70점 이상인 점수의 개수가 몇 개인지 계산한다.

```python
count = 0

for score in scores:
    if score >= 70:
        count += 1

print(count)
```

### Part 2. 딕셔너리를 이용한 단어 빈도 분석

>1. `split()`으로 문장을 단어 단위로 나눕니다.
>2. 빈 딕셔너리를 만듭니다.
>3. 단어를 하나씩 반복합니다.
>4. 처음 등장한 단어는 딕셔너리에 `1`로 저장합니다.
>5. 이미 존재하는 단어는 기존 값에 `1`을 증가시킵니다.
>6. 전체 단어와 등장 횟수를 출력합니다.
>7. 일정 횟수 이상 등장한 단어만 별도로 출력합니다. 

**필수 조건**

- `for` 사용
- `if` 사용
- `dict` 사용
- `collections.Counter` 등 자동 빈도 계산 기능은 사용하지 않음

**필수 역량**

`str`, `split()`, `dict`, `for`, `if`

```python
sentence="""
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""
```

1. `split()`으로 문장을 단어 단위로 나눕니다.

```python
words = sentence.split()
print(words)
```

2. 빈 딕셔너리를 만든다. 

```python
word_count = {}
```

3. 단어를 하나씩 반복한다.  
4. 처음 등장한 단어는 딕셔너리에 `1`로 저장합니다.  
5. 이미 존재하는 단어는 기존 값에 `1`을 증가시킵니다.

```python
for word in words:
    if word not in word_count:
        word_count[word] += 1

print(word_count)
```

6. 전체 단어와 등장 횟수를 출력합니다.

```python
for word in word_count:
    print(f"{word}: {word_count[word]}")
```

7. 일정 횟수 이상 등장한 단어만 별도로 출력합니다.

```python
for word in word_count:
    if word_count[word] > 1:
        print(f"{word}: {word_count[word]}")
    
```

### Part 3. 함수로 반복되는 코드 정리하기

```python
def count_words(sentence):
    words = sentence.split()
    word_count = {}
    
    for word in words:
        if word not in word_count:
            word_count[word] = 1
        else:
            word_count[word] += 1
            
    return word_count
```

```python
print(count_words(sentence))
```

```python
def count_words(sentence):
    words = sentence.split()
    word_count = {}

    for word in words:
        if word not in word_count:
            word_count[word] = 1
        else:
            word_count[word] += 1

    for word in word_count:
        if word_count[word] > 1:
            print(word, ":", word_count[word], "회")

    return word_count
```

```python
count_words(sentence)
```
