```python
#두 개의 숫자를 입력 받아서 더한 값을 반환하는 함수

#함수 이름 : first_sum

def first_sum(score1,score2):
    # result = score1+score2
    return score1+score2

num1=int(input("첫 번째 숫자:"))
num2=int(input("두 번째 숫자:"))

result=first_sum(num1, num2)

print("결과:", result)
```

> - 과제. (1~9)숫를 입력 받아서 사칙연산 결과를 반환하는 calc 라는 함수를 만드시오.
> 1. 4개 결과값을 한번에 반환 (Tuple)
> 2. 사용자가 입력한 값이 숫자가 아닐경우 오류메시지 출력하고 프로그램 종료

```python
def calc(num1,num2):
    sum = num1 + num2
    minus = num1 - num2
    multi = num1 * num2
    div = num1 / num2

    return sum, minus, multi, div

first = input(f"첫 번째 숫자:")
second = input(f"두 번재 숫자:")

lenth = len(first) + len(second)

if (lenth == 2 and ord(first) >= 49 and ord(first) <= 57) and (lenth == 2 and ord(second) >= 49 and ord(second) <= 57):
    #제대로 둘 다 숫자로 변환 가능
    result = calc (int(first), int(second))
    print(type(result))
    print(calc(1,2))
# result = calc(1,2)
else:
    #잘못된 값이 입련된 경우
    print("입력한 값(숫자)을 확인하세요!")
```
