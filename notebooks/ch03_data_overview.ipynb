```python
# 62. 반복 점검 함수

def inspect_dataframe(
    name: str,
    df: pd.DataFrame,
    primary_key: str | None = None,
) -> dict:
    result = {
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "column_names": df.columns.tolist(),
        "dtypes": {
            column: str(dtype)
            for column, dtype in df.dtypes.items()
        },
        "missing_total": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum()),
    }
    if primary_key is not None:
        result["primary_key"] = primary_key
        result["key_missing"] = int(
            df[primary_key].isna().sum()
        )
        result["key_duplicates"] = int(
            df[primary_key].duplicated().sum()
        )
        result["key_unique"] = int(
            df[primary_key].nunique()
        )
    return result
```

```python
# 62. 반복 점검 함수

def inspect_dataframe(
    name: str,
    df: pd.DataFrame,
    primary_key: str | None = None,
) -> dict:
    result = {
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "column_names": df.columns.tolist(),
        "dtypes": {
            column: str(dtype)
            for column, dtype in df.dtypes.items()
        },
        "missing_total": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum()),
    }
    if primary_key is not None:
        result["primary_key"] = primary_key
        result["key_missing"] = int(
            df[primary_key].isna().sum()
        )
        result["key_duplicates"] = int(
            df[primary_key].duplicated().sum()
        )
        result["key_unique"] = int(
            df[primary_key].nunique()
        )
    return result
```

## 10. 발견한 문제

1. `order_items`의 주요 ID(`order_item_id`)에 중복 1건 존재 (key_duplicates=1)
2. `orders.customer_id`가 참조하는 `customers.customer_id` 중 존재하지 않는 값 1건
3. `order_items.order_id`가 참조하는 `orders.order_id` 중 존재하지 않는 값 1건
4. `order_items.product_id`가 참조하는 `products.product_id` 중 존재하지 않는 값 1건

## 11. 즉시 처리하지 않은 이유

- 이 요약 정보만으로는 **어떤 행(row)**이 문제인지, 구체적으로 어떤 값이 중복·불일치하는지 알 수 없음
- 원인 파악(입력 오류인지, 참조 대상 테이블의 누락인지) 없이 건수만 보고 삭제·수정하면 실제 데이터를 잘못 훼손할 위험이 있음
- 결측치가 아니라 **중복/참조 오류**이므로 일반적인 결측치 처리 방식(삭제 등)을 그대로 적용할 수 없음

## 12. 다음 전처리 단계에서 결정할 내용

- 원본 파일을 직접 열어 `order_items` 중복 ID(1건)의 실제 행 2개를 비교해 어느 것이 올바른 데이터인지 결정
- 참조무결성 위반 3건(주문↔고객, 주문상세↔주문, 주문상세↔상품)의 실제 값을 확인해 원인(오탈자 vs 마스터 테이블 누락)을 파악
- 숫자형 범위·범주형 값·날짜 형식은 이번 요약에 없으므로, 별도로 원본 파일을 검사할지 여부 결정
- 위 문제 행들을 분석에 포함할지, 제외할지, 별도 플래그로 표시할지 결정
