핵심 용어 한 줄 정리

• Rule-based: 사람이 규칙을 직접 작성
• Machine Learning: 데이터에서 판단 패턴을 학습
• Deep Learning: 복잡한 특징까지 모델이 더 많이 학습
• NLP: 사람의 언어를 컴퓨터가 처리하는 분야
• Tokenization: 문장을 처리 가능한 작은 단위로 나누는 과정
• Vectorization: 데이터를 숫자 벡터로 표현하는 넓은 개념
• Embedding: 의미가 반영되도록 학습된 벡터 표현
• Similarity: 두 벡터의 의미적 가까움을 계산
• Transformer: 문맥과 단어 간 관계를 효과적으로 처리하는 모델 구조
• LLM: 다양한 언어 업무를 수행하는 대규모 언어 모델
• RAG: 필요한 자료를 검색해 LLM 답변에 활용
• Agent: 목표를 받아 도구를 사용하며 실제 업무를 수행
• MCP: AI와 외부 도구·데이터 연결 방식을 표준화하려는 프로토콜
• Multi-Agent: 여러 Agent가 역할을 나누어 협업하는 구조
