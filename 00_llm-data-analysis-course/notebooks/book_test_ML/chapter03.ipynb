```python

import pandas as pd

DATA_PATH = "book_bestseller_clean.csv"
df_books = pd.read_csv(DATA_PATH, encoding="utf-8-sig")

print("데이터 크기:", df_books.shape)
print("컬럼:", df_books.columns.tolist())
print("상품명 결측치:", df_books["상품명"].isna().sum())
df_books[["상품명"]].head(10)
```

```python
titles = (
    df_books["상품명"]
    .fillna("")
    .astype(str)
    .str.strip()
)

titles = titles[titles != ""].reset_index(drop=True)
print("사용할 제목 수:", len(titles))
titles.head(10)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# 1. 예시 문서 정의
docs = [
    "파이썬 데이터 분석 책 추천",
    "딥러닝 인공지능 도서 추천",
    "소설 에세이 베스트셀러 책 도서 리뷰"
]

# 2. Count 방식
count_vec = CountVectorizer(token_pattern=r"(?u)\b\w+\b")
count_matrix = count_vec.fit_transform(docs)
words = count_vec.get_feature_names_out()

df_count = pd.DataFrame(count_matrix.toarray(), columns=words)
print("=== [1] CountVectorizer 결과 ===")
print(df_count)

print("-" * 50)

# 3. TF-IDF 방식
tfidf_vec = TfidfVectorizer(token_pattern=r"(?u)\b\w+\b")
tfidf_matrix = tfidf_vec.fit_transform(docs)

# 이 부분의 주석(#)이 완전히 해제되어 있어야 에러가 나지 않습니다.
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=words)
print("=== [2] TfidfVectorizer 결과 ===")
print(df_tfidf.round(2))
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# 1. 이미지에 나온 예시 문서 정의 (파이썬 2번, 데이터 1번, 분석 1번)
docs = [
    "파이썬 파이썬 데이터 분석"
]

# 2. 순수한 등장 횟수(TF) 확인을 위한 CountVectorizer
count_vec = CountVectorizer()
count_matrix = count_vec.fit_transform(docs)
words = count_vec.get_feature_names_out()

df_count = pd.DataFrame(count_matrix.toarray(), columns=words)
print("=== [1] 단순 등장 횟수 (Count) ===")
print(df_count)
print("-" * 50)

# 3. scikit-learn의 정규화가 적용된 TfidfVectorizer
tfidf_vec = TfidfVectorizer()
tfidf_matrix = tfidf_vec.fit_transform(docs)

df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=words)
print("=== [2] scikit-learn의 최종 TF-IDF 값 ===")
print(df_tfidf.round(2))
```
